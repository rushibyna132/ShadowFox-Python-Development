```python

import random

count_6 = 0
count_1 = 0
two_sixes = 0
previous_roll = 0

for i in range(20):
    roll = random.randint(1, 6)
    print("Roll", i + 1, ":", roll)

    if roll == 6:
        count_6 += 1

    if roll == 1:
        count_1 += 1

    if roll == 6 and previous_roll == 6:
        two_sixes += 1

    previous_roll = roll

print("Number of 6s:", count_6)
print("Number of 1s:", count_1)
print("Two 6s in a row:", two_sixes)
```

```python
total = 0

for i in range(10):
    total += 10
    print("Total jumping jacks:", total)

    if total == 100:
        print("Congratulations! You completed the workout.")
        break

    tired = input("Are you tired? (yes/no): ").lower()

    if tired == "yes" or tired == "y":
        skip = input("Do you want to skip the remaining sets? (yes/no): ").lower()

        if skip == "yes" or skip == "y":
            print("You completed a total of", total, "jumping jacks.")
            break
        else:
            print("Remaining jumping jacks:", 100 - total)
    else:
        print("Remaining jumping jacks:", 100 - total)
```

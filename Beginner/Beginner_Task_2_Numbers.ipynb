```python
# Question 1: Number Formatting

def number_format(num, form):
    result = format(num, form)
    return result

answer = number_format(145, 'o')
print("Octal representation:", answer)
```

```python
# Question 2: Area of Circular Pond

pi = 3.14
r = 84

area = pi * r * r

print("Area of pond:", area)
```

```python
# Bonus: Calculate Total Water

water = area * 1.4

print("Total water:", round(water))
```

```python
# Question 3: Calculate Speed

distance = 490
time = 7 * 60

speed = distance / time

print("Speed:", round(speed), "m/s")
```

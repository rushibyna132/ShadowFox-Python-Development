```python
# Import required libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris

# Load the Iris dataset

iris = load_iris(as_frame=True)

df = iris.frame

# Add flower species names

df["species"] = df["target"].map(
    dict(enumerate(iris.target_names))
)

# Remove the target column

df = df.drop("target", axis=1)

# Display first five rows

print(df.head())
```

```python
# Display dataset information

print("Dataset Information:")
df.info()

# Check missing values

print("\nMissing Values:")
print(df.isnull().sum())

# Display dataset shape

print("\nDataset Shape:")
print(df.shape)
```

```python
# Statistical summary

print("Statistical Summary:")
print(df.describe())
```

```python
# Count the number of flowers in each species

print("Species Count:")
print(df["species"].value_counts())
```

```python
# Count plot of flower species

sns.countplot(data=df, x="species")

plt.title("Distribution of Iris Flower Species")
plt.xlabel("Flower Species")
plt.ylabel("Number of Flowers")

plt.show()
```

```python
# Compare petal lengths of different species

sns.boxplot(data=df, x="species", y="petal length (cm)")

plt.title("Petal Length Comparison")
plt.xlabel("Flower Species")
plt.ylabel("Petal Length (cm)")

plt.show()
```

```python
# Scatter plot

sns.scatterplot(
    data=df,
    x="petal length (cm)",
    y="petal width (cm)",
    hue="species"
)

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")

plt.show()
```

```python
# Calculate correlation

correlation = df.select_dtypes(
    include="number"
).corr()

print("Correlation Matrix:")
print(correlation)

# Create heatmap

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
# Average measurements of each species

species_analysis = df.groupby("species").mean()

print("Average Measurements by Species:")
print(species_analysis)
```
